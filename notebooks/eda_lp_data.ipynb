# Análise Exploratória de Dados (EDA) - Modelo de Inferência do Perfil de Aprendizagem (LP)

Este notebook tem como objetivo realizar a Análise Exploratória de Dados (EDA) sobre o dataset simulado de perfis de aprendizagem. O objetivo é entender a estrutura dos dados, a distribuição das features e do target, e identificar possíveis padrões e insights iniciais.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Configurações para visualização
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["font.size"] = 12
```

## 1. Carregamento dos Dados

```python
data_path = os.path.join('..', 'data', 'raw', 'simulated_lp_data.csv')
df = pd.read_csv(data_path)

print(f"Dataset carregado com {df.shape[0]} linhas e {df.shape[1]} colunas.")
df.head()
```

## 2. Visão Geral dos Dados

```python
print("Informações do DataFrame:")
df.info()
```

```python
print("Estatísticas Descritivas:")
df.describe()
```

## 3. Análise do Target: `learning_profiles`

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='learning_profiles', palette='viridis')
plt.title('Distribuição dos Perfis de Aprendizagem')
plt.xlabel('Perfil de Aprendizagem')
plt.ylabel('Contagem')
plt.show()

print("Contagem de cada perfil de aprendizagem:")
print(df['learning_profiles'].value_counts(normalize=True) * 100)
```

## 4. Análise das Features Numéricas

```python
numerical_features = df.select_dtypes(include=np.number).columns.tolist()

for feature in numerical_features:
    plt.figure(figsize=(12, 6))
    sns.histplot(data=df, x=feature, kde=True, hue='learning_profiles', palette='viridis')
    plt.title(f'Distribuição de {feature} por Perfil de Aprendizagem')
    plt.xlabel(feature)
    plt.ylabel('Frequência')
    plt.show()

    plt.figure(figsize=(12, 6))
    sns.boxplot(data=df, x='learning_profiles', y=feature, palette='viridis')
    plt.title(f'Boxplot de {feature} por Perfil de Aprendizagem')
    plt.xlabel('Perfil de Aprendizagem')
    plt.ylabel(feature)
    plt.show()
```

## 5. Análise da Feature Categórica: `most_preferred_resource_type`

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='most_preferred_resource_type', hue='learning_profiles', palette='viridis')
plt.title('Preferência de Recurso por Perfil de Aprendizagem')
plt.xlabel('Tipo de Recurso Mais Preferido')
plt.ylabel('Contagem')
plt.xticks(rotation=45)
plt.legend(title='Perfil')
plt.tight_layout()
plt.show()
```

## 6. Matriz de Correlação (Features Numéricas)

```python
plt.figure(figsize=(12, 10))
correlation_matrix = df[numerical_features].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matriz de Correlação das Features Numéricas')
plt.show()
```
